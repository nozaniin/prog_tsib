Дан файл с результатами голосования на выборах в виде:

1 3 14 10 2 3 -1 ...

где номер (через пробел) определяет голос за партию из списка:
1. Партия №1.
2. Партия №2.
3. Партия №3.
4. Партия №4.
5. Партия №5.

  -1. Испорченный бланк.

Если при чтении файла какое-либо значение не распознается (не является целым
числом) или партии с таким номером не существует, считается, что бланк испорчен
(-1).

Подведите итоги выборов, выведя на экран список партий в соответствии с
убыванием количества полученных голосов и их процентным соотношением:
1. Партия №2 | 1111 | 58.21%
2. Партия №4 | 999 | 38.14%

...


```python
CORRUPTED = -1
parties = {
    1: "Партия №1", 2: "Партия №2", 3: "Партия №3",
    4: "Партия №4", 5: "Партия №5",
    CORRUPTED: "Испорчено"}     #создаем словарь с партиями

def load_votes(filename):    #загружаем голоса из файла
    votes = []
    with open(filename, 'r', encoding='utf-8') as f:    #открываем файл
        for line in f:    # для каждой строки
            for part in line.split():    # для каждого числа в строке
                try:
                    votes.append(int(part))
                except ValueError:
                    votes.append(CORRUPTED)
    return votes   # возвращаем словарь с голосами

def parties_votes(parties, votes):    # подсчитываем голоса по партиям
    result = {name: 0 for name in parties.values()}   # начальные нули
    for vote in votes:
        if vote in parties:
            result[parties[vote]] += 1    #подсчитываем голоса, если есть совпадение со словарем
        else:
            result[parties[CORRUPTED]] += 1
    return result

filename = "votes.txt"
votes = load_votes(filename)    #загружаем голоса
stats = parties_votes(parties, votes)    # подсчитываем голоса
total = len(votes)

for name, count in sorted(stats.items(), key=lambda x: x[1], reverse=True):    # сортируем по голосам (убывание)
    if name != parties[CORRUPTED]:    # не выводим "Испорчено"
        percent = count / total * 100
        print(f"{name} | {count} | {percent:.2f}%")
```
