3. Сравните способы решения, сопоставив их асимптотические и временные
сложности (устно).

```python
# Первый способ - наиболее оптимальный по сложности
def foo(nums):
    """Вычисляет сумму первого элемента и квадрата последнего элемента списка
    Параметры:
        - nums (list): список чисел (предполагается, что список не пуст и содержит хотя бы один элемент)
    Сложность: O(1) – время выполнения не зависит от длины списка, так как доступ к элементам по индексу и арифметические операции выполняются за константное время
    """
    return (nums[0] + nums[-1] ** 2)


# Второй способ - менее оптимальный по сложности
def foo_2(nums):
    """Альтернатива прошлому решению – явный цикл
    Параметры:
        - nums (list): список чисел
    Сложность: O(n) – один проход по списку для поиска последнего элемента
    """
    first = nums[0]
    last = None
    for elem in nums:   # цикл всё равно пройдёт все элементы, но last станет последним
        last = elem
    return first + last ** 2
```
