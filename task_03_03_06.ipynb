```python
"""
Ошибки (номера строк через пробел, данная строка - №2): 23 27 28 29
"""

def unemployment_rate(unemployed, employed):
    """Вернуть уровень безработицы (УБ) в долях 1.
       Расчет по формуле: УБ = Безработные / (Занятые + Безработные).
       Параметры:
       - unemployed, employed (int или float) — неотрицательные числа.
       Результат:
       - float.
       Исключения:
       - ValueError: если хотя бы один аргумент отрицателен или сумма равна нулю.
    """
    if unemployed < 0 or employed < 0:
        raise ValueError("Количество безработных и занятых не может быть отрицательным.")
    if unemployed + employed == 0:
        raise ValueError("Общая численность рабочей силы не может быть нулевой.")
    return unemployed / (unemployed + employed)


try:
    unemployed = int(input("Введите кол-во безработных (чел.): "))
    employed = int(input("Введите кол-во занятых (чел.): "))
    rate = unemployment_rate(unemployed, employed)
    print("Уровень безработицы = {:.1%}".format(rate))
except ValueError as err:
    print("Ошибка:", err, "Проверьте введенные числа.")
except Exception as err:
    print("Непредвиденная ошибка:", err)
```
